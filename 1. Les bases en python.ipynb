## 1.4 Quelques calculs élémentaires

```python
a = 10
b = 7

print(a + b)   # addition
print(a - b)   # soustraction
print(a * b)   # multiplication
print(a / b)   # division
print(a // b)  # division entière
print(a % b)   # reste
print(a ** b)  # puissance
```

## 1.5 La structure conditionnelle

```python
prix_tableau_euros = 25000
budget_galerie_euros = 30000

if budget_galerie_euros >= prix_tableau_euros:
    print("La galerie peut acheter le tableau.")
else:
    print("La galerie n'a pas assez d'argent pour acheter le tableau.")
```

```python
oeuvre_est_authentique = True

if oeuvre_est_authentique == True:
    print("Le musée peut exposer cette œuvre.")
else:
    print("Attention : l'œuvre doit être authentifiée.")
```

On peut ajouter plusieurs confitions avec `elif`.

```python
annee = 1550

if annee < -3000:
    print("Préhistoire")
elif annee < 476:
    print("Antiquité")
elif annee < 1492:
    print("Moyen Âge")
elif annee < 1789:
    print("Époque moderne")
else:
    print("Époque contemporaine")
```

## 1.6 Les boucles

La boucle ``for`` répète une action plusieurs fois.

``range(n)`` génère les nombres de 0 à n-1, qui sont affichés un par un.

```python
for nombre in range(5):
    print(nombre)
```
